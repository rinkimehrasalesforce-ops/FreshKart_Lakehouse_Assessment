```python
dbutils.widgets.text("catalog", "fk_ce1366")
catalog = dbutils.widgets.get("catalog")
```

```python
from pyspark.sql.functions import *
```

```python
df = spark.read.table(f"{catalog}.bronze.stores")

#trim and initcap
df = (
    df.filter(col("store_id").isNotNull())
    .withColumn("store_name", trim(initcap(col("store_name"))))
    .withColumn("city", trim(initcap(col("city"))))
    .withColumn("state", trim(initcap(col("state"))))
    .withColumn("region", trim(initcap(col("region"))))
    .withColumn("opening_date", to_date(col("opening_date")))
)

df = (
    df.drop("_rescued_data")
    .drop("ingested_at")
    .drop("_source_file")
)

df.write.mode("overwrite").saveAsTable(f"{catalog}.silver.stores")
```
